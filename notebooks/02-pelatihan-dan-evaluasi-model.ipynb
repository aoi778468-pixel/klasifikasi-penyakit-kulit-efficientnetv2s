```python
"""Skin Disease Classifier - 6 kelas
EfficientNetV2-S 384px | balanced sampling | MixUp | cosine LR | TTA 4 view | ensemble
"""
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.applications import EfficientNetV2S
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import AdamW
from sklearn.metrics import classification_report, confusion_matrix

# ===== Konfigurasi =====
BASE_DIR = "/kaggle/input/notebooks/aoisaputra/fildat-2/dataset_split"
IMG_SIZE = (384, 384)
BATCH_SIZE = 16
NUM_MODELS = 2              # jumlah model ensemble (1 = tanpa ensemble)
BASE_SEED = 42
EPOCH_HEAD, EPOCH_FINE = 6, 30
LR_HEAD, LR_FINE = 1e-3, 1e-4
MIXUP_ALPHA = 0.2           # 0 = matikan MixUp
LABEL_SMOOTHING = 0.1

AUTOTUNE = tf.data.AUTOTUNE
IMG_EXTS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

print("GPU:", tf.config.list_physical_devices("GPU"))


# ===== Data =====
def load_eval_ds(split):
    return tf.keras.utils.image_dataset_from_directory(
        os.path.join(BASE_DIR, split), image_size=IMG_SIZE, batch_size=BATCH_SIZE,
        shuffle=False, label_mode="categorical")


val_ds, test_ds = load_eval_ds("val"), load_eval_ds("test")
CLASS_NAMES = val_ds.class_names
NUM_CLASSES = len(CLASS_NAMES)
val_ds, test_ds = val_ds.prefetch(AUTOTUNE), test_ds.prefetch(AUTOTUNE)
print("Kelas:", CLASS_NAMES)

# Daftar file train per kelas
train_files = []
for name in CLASS_NAMES:
    folder = os.path.join(BASE_DIR, "train", name)
    train_files.append([os.path.join(folder, f) for f in os.listdir(folder)
                        if f.lower().endswith(IMG_EXTS)])
counts = np.array([len(f) for f in train_files], dtype=float)
print("Train per kelas:", dict(zip(CLASS_NAMES, counts.astype(int))))

# Bobot sampling ~ sqrt(jumlah): kelas kecil lebih sering muncul tanpa dipaksa seimbang penuh
SAMPLE_WEIGHTS = list(np.sqrt(counts) / np.sqrt(counts).sum())
STEPS_PER_EPOCH = int(counts.sum() // BATCH_SIZE)


def load_image(path, label):
    img = tf.io.decode_image(tf.io.read_file(path), channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE)
    img.set_shape(IMG_SIZE + (3,))
    return img, tf.one_hot(label, NUM_CLASSES)


def mixup(x, y):
    n = tf.shape(x)[0]
    g1, g2 = tf.random.gamma([n], MIXUP_ALPHA), tf.random.gamma([n], MIXUP_ALPHA)
    lam = g1 / (g1 + g2 + 1e-8)
    lx, ly = tf.reshape(lam, [-1, 1, 1, 1]), tf.reshape(lam, [-1, 1])
    return (lx * x + (1 - lx) * tf.reverse(x, [0]),
            ly * y + (1 - ly) * tf.reverse(y, [0]))


def build_train_ds(seed, use_mixup):
    per_class = [
        tf.data.Dataset.from_tensor_slices((files, [i] * len(files)))
        .shuffle(len(files), seed=seed).repeat()
        .map(load_image, num_parallel_calls=AUTOTUNE)
        for i, files in enumerate(train_files)
    ]
    ds = tf.data.Dataset.sample_from_datasets(per_class, weights=SAMPLE_WEIGHTS, seed=seed)
    ds = ds.batch(BATCH_SIZE, drop_remainder=True)
    if use_mixup:
        ds = ds.map(mixup, num_parallel_calls=AUTOTUNE)
    return ds.prefetch(AUTOTUNE)


# ===== Model =====
def build_model(seed):
    augment = tf.keras.Sequential([
        layers.RandomFlip("horizontal_and_vertical", seed=seed),
        layers.RandomRotation(0.15, seed=seed),
        layers.RandomZoom((-0.2, 0.1), seed=seed),
        layers.RandomTranslation(0.1, 0.1, seed=seed),
        layers.RandomContrast(0.2, seed=seed),
        layers.RandomBrightness(0.2, value_range=(0, 255), seed=seed),
    ])
    base = EfficientNetV2S(include_top=False, weights="imagenet",
                           input_shape=IMG_SIZE + (3,))
    base.trainable = False

    inp = layers.Input(shape=IMG_SIZE + (3,))
    x = augment(inp)
    x = tf.keras.applications.efficientnet_v2.preprocess_input(x)
    x = base(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dense(256, activation="relu", kernel_regularizer=regularizers.l2(1e-4))(x)
    x = layers.Dropout(0.4)(x)
    out = layers.Dense(NUM_CLASSES)(x)  # logits
    return models.Model(inp, out), base


def compile_model(model, lr):
    model.compile(
        optimizer=AdamW(lr, weight_decay=1e-4, clipnorm=1.0),
        loss=tf.keras.losses.CategoricalCrossentropy(
            from_logits=True, label_smoothing=LABEL_SMOOTHING),
        metrics=["accuracy"])


def train_one(seed):
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)
    model, base = build_model(seed)

    # Tahap 1: latih head saja
    compile_model(model, LR_HEAD)
    h1 = model.fit(build_train_ds(seed, use_mixup=False), validation_data=val_ds,
                   epochs=EPOCH_HEAD, steps_per_epoch=STEPS_PER_EPOCH, verbose=2)

    # Tahap 2: fine-tune semua layer (BatchNorm tetap beku), cosine decay + MixUp
    for layer in base.layers:
        if not isinstance(layer, layers.BatchNormalization):
            layer.trainable = True
    compile_model(model, tf.keras.optimizers.schedules.CosineDecay(
        LR_FINE, EPOCH_FINE * STEPS_PER_EPOCH, alpha=0.01))
    h2 = model.fit(
        build_train_ds(seed, use_mixup=MIXUP_ALPHA > 0), validation_data=val_ds,
        epochs=EPOCH_FINE, steps_per_epoch=STEPS_PER_EPOCH, verbose=2,
        callbacks=[EarlyStopping(monitor="val_accuracy", mode="max",
                                 patience=8, restore_best_weights=True)])

    history = {k: h1.history[k] + h2.history[k]
               for k in ("accuracy", "val_accuracy", "loss", "val_loss")}
    return model, history


# ===== Evaluasi: TTA 4 view (asli, flip H, flip V, flip H+V) =====
TTA_VIEWS = [
    lambda x: x,
    tf.image.flip_left_right,
    tf.image.flip_up_down,
    lambda x: tf.image.flip_up_down(tf.image.flip_left_right(x)),
]


def predict_tta(model, ds):
    probs = []
    for view in TTA_VIEWS:
        d = ds.map(lambda x, y, v=view: (v(x), y), num_parallel_calls=AUTOTUNE)
        probs.append(tf.nn.softmax(model.predict(d, verbose=0), axis=1).numpy())
    return np.mean(probs, axis=0)


def report(name, y_true, y_pred):
    print(f"{name}: accuracy={np.mean(y_true == y_pred):.4f}")


def plot_history(h):
    epochs = range(1, len(h["accuracy"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    for ax, title, k in [(axes[0], "Accuracy", "accuracy"), (axes[1], "Loss", "loss")]:
        ax.plot(epochs, h[k], color="tab:red", label="train")
        ax.plot(epochs, h["val_" + k], color="tab:blue", label="val")
        ax.axvline(EPOCH_HEAD, ls="--", c="gray")
        ax.set(title=f"Model {title}", xlabel="Epoch")
        ax.legend()
    plt.tight_layout()
    plt.savefig("training_history.png", dpi=150)
    plt.show()


def plot_confusion(y_true, y_pred):
    plt.figure(figsize=(8, 6))
    sns.heatmap(confusion_matrix(y_true, y_pred), annot=True, fmt="d", cmap="Blues",
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.savefig("confusion_matrix.png", dpi=150)
    plt.show()


# ===== Main =====
y_true = np.concatenate([np.argmax(y.numpy(), axis=1) for _, y in test_ds])
all_probs = []

for k in range(NUM_MODELS):
    seed = BASE_SEED + k * 17
    print(f"\n===== Model {k + 1}/{NUM_MODELS} (seed={seed}) =====")
    model, history = train_one(seed)
    model.save(f"model_seed{seed}.keras")

    probs = predict_tta(model, test_ds)
    report(f"Model {k + 1} + TTA", y_true, probs.argmax(1))
    all_probs.append(probs)

y_pred = np.mean(all_probs, axis=0).argmax(1)
print()
report(f"ENSEMBLE {NUM_MODELS} model + TTA", y_true, y_pred)
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES,
                            digits=2, zero_division=0))

plot_history(history)  # riwayat model terakhir
plot_confusion(y_true, y_pred)
```
